Мои предсказания и объяснения:

1. 250001.replace(" ", "") убирает пробел, получается строка "25000". Дальше + "1" склеивает две строки, а не складывает числа, поэтому результат "250001", а не 25001.
2. almaty.strip() убирает пробелы по краям, а .lower() переводит все буквы в нижний регистр.
3. 9 (3, 4)np.arange(12) даёт числа от 0 до 11. .reshape(3, 4) раскладывает их в таблицу из 3 строк и 4 столбцов.  a[2, 1] — это строка с индексом 2 и столбец с индексом 1 (счёт идёт с нуля), то есть число 9. a.shape равно (3, 4).
4. 1 [9 2] argmax() возвращает индекс, а не само значение. Максимум 9 встречается дважды, и берётся первое вхождение, у которого индекс 1. Срез a[1:3] включает элементы с индексами 1 и 2, а элемент с индексом 3 не входит.
5. (2, 3) [10 7 4 1] np.zeros((2, 3)) создаёт массив из нулей формы 2×3, поэтому .shape равно (2, 3). np.arange(10, 0, -3) начинает с 10 и уменьшает на 3: 10, 7, 4, 1. Конечное значение 0 в результат не входит.

### Задача 4. Заказы в NumPy
Соберите из `rows` массивы `price`, `qty`, `product` (цикл для сбора списков можно, как на уроке). Посчитайте `revenue = price * qty`.
1. Сколько всего заказов?
2. Самый **дешёвый** заказ: сумма и товар (`argmin`).
3. Самый дорогой заказ: сумма. Во сколько раз он дороже самого дешёвого?

**Пример проверки:** `500` заказов; самый дешёвый — `7 500 ₸`, `Мышь Logitech M185`; самый дорогой — `1 098 000 ₸`.

````python
price, qty, product = [], [], []
for r in rows:
    price.append(int(float(r["price"])))
    qty.append(int(r["qty"]))
    product.append(r["product"])

price = np.array(price)
qty = np.array(qty)
product = np.array(product)

revenue = price * qty

# 1) сколько заказов
print("Заказов:", len(revenue))

# 2) самый дешёвый заказ
i_min = revenue.argmin()
print("Самый дешёвый:", f"{revenue[i_min]:,}".replace(",", " "), "₸,", product[i_min])

# 3) самый дорогой заказ
i_max = revenue.argmax()
print("Самый дорогой:", f"{revenue[i_max]:,}".replace(",", " "), "₸")
print("Дороже в", round(revenue[i_max] / revenue[i_min], 1), "раз")
````

### Задача 6. Сколько потеряли на отменах
Соберите массив `status`. Посчитайте:
- сколько заказов отменено;
- какая это доля от всех заказов;
- на какую сумму были отменённые заказы.

Подсказка: `status == "отменён"` — массив `True/False`; `.sum()` считает `True`, `.mean()` даёт долю.

**Пример проверки:** `45` заказов, `9%`, `8 224 000 ₸`.

````python
status = np.array([r["status"].strip() for r in rows])

cancelled = status == "отменён"          # массив True/False

n_cancel = cancelled.sum()
share = cancelled.mean()
lost = revenue[cancelled].sum()

print("Отменено:", n_cancel)
print("Доля:", f"{share:.0%}")
print("Потеряли:", f"{lost:,}".replace(",", " "), "₸")
````

### Задача 7. Неделя курьера
Курьер развозит по 5 заказов в день. Возьмите первые **35** заказов и разложите в таблицу «дни × заказы».
1. Какую форму передать в `reshape`?
2. Посчитайте сумму за каждый день — список из 7 чисел (цикл по строкам таблицы можно).
3. В какой день сумма была самой большой?

**Пример проверки:** самый «дорогой» день — **7-й**, `1 227 000 ₸`.

````python
week = revenue[:35].reshape(7, 5)     # 1) 7 дней × 5 заказов
print(week.shape)

# 2) сумма за каждый день
day_sums = []
for day in week:
    day_sums.append(int(day.sum()))
print(day_sums)

# 3) самый «дорогой» день
best = np.argmax(day_sums)
print("День:", best + 1, "—", f"{day_sums[best]:,}".replace(",", " "), "₸")
````

### Задача 8. Средний чек по городам
Соберите массив `city`. Для **каждого** города из `np.unique(city)` посчитайте число заказов и средний чек
(цикл по городам можно, внутри — маски). Сохраните в `cities.json`:
```json
{
  "Актобе": {"заказов": 33, "средний_чек": 137136},
  "Алматы": {"заказов": 133, "средний_чек": 193011},
  ...
}
```
В каком городе самый высокий средний чек? Почему в Атырау средний чек считать рискованно?

**Пример проверки:** самый высокий средний чек — **Караганда**, ≈ `208 826 ₸`.

````python
city = np.array([r["city"].strip() for r in rows])

cities = {}
for c in np.unique(city):
    mask = city == c
    cities[str(c)] = {
        "заказов": int(mask.sum()),
        "средний_чек": int(round(revenue[mask].mean())),
    }

with open("cities.json", "w", encoding="utf-8") as f:
    json.dump(cities, f, ensure_ascii=False, indent=2)

for c, d in cities.items():
    print(c, d)

top = max(cities, key=lambda c: cities[c]["средний_чек"])
print("Самый высокий средний чек:", top, cities[top]["средний_чек"])
````
